# Collecting X and Y

```python
X=data['MinTemp'].values
Y=data['MaxTemp'].values
```

# Find the mean of X and the mean of Y

```python
mean_X=np.mean(X)
mean_Y=np.mean(Y)
#print(mean_X)
#print(mean_Y)
```

# Total number of values

```python
n=len(X)
#n
```

# Using the formula to calculate b0 and b1

```python
numerator=0
denominator=0

for i in range(n):
    numerator +=(X[i]-mean_X)*(Y[i]-mean_Y)
    denominator +=(X[i]-mean_X)**2
    
b1=numerator/denominator
b0=mean_Y-(b1*mean_X)
```

# Print coefficients

```python
print(b1,b0)
```

# Plotting values and drawing regression line

```python
max_X= np.max(X)+10
min_X=np.min(X)-10
```

# Calculating line values of X and Y

```python
x=np.linspace(min_X,max_X,300)
y=b0+b1*x
#print(x)
```

# Plotting line and Scatter points

```python
plt.plot(x,y,color='red',label='Regression Line')
plt.scatter(X,Y,color='black',label='Scatter Plot')
plt.xlabel('MinTemp')
plt.ylabel('MaxTemp')
plt.legend()
plt.show()
```

# Calculating R squared

```python
ss_t=0
ss_r=0

for i in range(n):
    y_pred=b0+b1*X[i]
    ss_t +=(Y[i]-mean_Y)**2
    ss_r +=(Y[i]-y_pred)**2

r2=1-(ss_r/ss_t)    
```

```python
print(r2)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```
